2.   Altere o valor do parâmetro de regularização `C` para 5 valores onde $0.1<C<100$. compare e comente os resultados.

```python
# Exercício 1.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

X, y = make_moons(n_samples=600, noise=0.15, random_state=0)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=0
)

C = 10

modelo_linear = SVC(kernel='linear', C=C)
modelo_linear.fit(X_train, y_train)
acuracia_linear = modelo_linear.score(X_test, y_test)

modelo_poly = SVC(kernel='poly', C=C)
modelo_poly.fit(X_train, y_train)
acuracia_poly = modelo_poly.score(X_test, y_test)

modelo_sigmoid = SVC(kernel='sigmoid', C=C)
modelo_sigmoid.fit(X_train, y_train)
acuracia_sigmoid = modelo_sigmoid.score(X_test, y_test)

print("Acurácia Linear:", acuracia_linear)
print("Acurácia Polinomial:", acuracia_poly)
print("Acurácia Sigmoid:", acuracia_sigmoid)
```

Comentário:

O kernel linear apresentou um resultado inferior porque o conjunto make_moons possui um formato que não pode ser separado facilmente por uma linha reta.

O kernel polinomial conseguiu representar melhor a separação entre as classes, apresentando uma melhora na classificação.

Já o sigmoid também consegue criar uma separação não linear, mas neste caso apresentou um resultado inferior ao polinomial.

De forma geral, para o conjunto make_moons, os kernels não lineares tendem a apresentar melhores resultados que o linear.

```python
# Exercício 2.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

X, y = make_moons(n_samples=600, noise=0.15, random_state=0)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=0
)

valores_C = [0.1, 1, 10, 50, 99]

for C in valores_C:

    modelo = SVC(kernel='rbf', C=C)
    modelo.fit(X_train, y_train)

    acuracia = modelo.score(X_test, y_test)

    print("C =", C, "- Acurácia =", acuracia)
```

Foi possível observar que a alteração do valor de C influencia diretamente o resultado do SVM.

Com valores muito baixos, como C = 0.1, o modelo possui uma margem maior e aceita mais erros na classificação.

Conforme o valor de C aumenta, o modelo passa a tentar classificar os dados de treinamento com maior precisão.

Porém, aumentar muito o valor de C não significa necessariamente obter um modelo melhor, pois existe a possibilidade de ocorrer overfitting.

Nesse conjunto de dados, os valores intermediários e altos de C apresentaram os melhores resultados.

```python
# Exercício 3.

import numpy as np
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

valores_noise = [0.05, 0.10, 0.15, 0.25, 0.35]

for noise in valores_noise:

    X, y = make_moons(
        n_samples=600,
        noise=noise,
        random_state=0
    )

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=0
    )

    modelo = SVC(kernel='rbf', C=10)

    modelo.fit(X_train, y_train)

    acuracia = modelo.score(X_test, y_test)

    print("Noise =", noise, "- Acurácia =", acuracia)
```

O parâmetro noise adiciona pequenas alterações nos pontos do conjunto de dados.

Quando utilizamos um valor baixo de noise, os pontos ficam mais organizados e separados, facilitando o trabalho do SVM.

À medida que aumentamos o noise, os pontos ficam mais espalhados e algumas classes começam a ficar mais próximas umas das outras.

Por isso, de forma geral, quanto maior o ruído, maior a dificuldade para o modelo realizar a classificação corretamente.

O kernel rbf apresentou bons resultados porque consegue criar uma separação não linear entre as duas classes.

```python
# Exercício 4.

import numpy as np
from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

X, y = make_circles(
    n_samples=500,
    noise=0.15,
    factor=0.4,
    random_state=0
)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=0
)

C = 10

modelo_linear = SVC(kernel='linear', C=C)
modelo_linear.fit(X_train, y_train)

modelo_poly = SVC(kernel='poly', C=C)
modelo_poly.fit(X_train, y_train)

modelo_sigmoid = SVC(kernel='sigmoid', C=C)
modelo_sigmoid.fit(X_train, y_train)

modelo_rbf = SVC(kernel='rbf', C=C)
modelo_rbf.fit(X_train, y_train)

print("Linear:", modelo_linear.score(X_test, y_test))
print("Polinomial:", modelo_poly.score(X_test, y_test))
print("Sigmoid:", modelo_sigmoid.score(X_test, y_test))
print("RBF:", modelo_rbf.score(X_test, y_test))


#Testando diferentes valores de C
valores_C = [0.1, 1, 10, 50, 99]

for C in valores_C:

    modelo = SVC(kernel='rbf', C=C)
    modelo.fit(X_train, y_train)

    acuracia = modelo.score(X_test, y_test)

    print("C =", C, "- Acurácia =", acuracia)


#Testando diferentes valores de noise
valores_noise = [0.05, 0.10, 0.15, 0.25, 0.35]

for noise in valores_noise:

    X, y = make_circles(
        n_samples=500,
        noise=noise,
        factor=0.4,
        random_state=0
    )

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=0
    )

    modelo = SVC(kernel='rbf', C=10)
    modelo.fit(X_train, y_train)

    acuracia = modelo.score(X_test, y_test)

    print("Noise =", noise, "- Acurácia =", acuracia)
```

O make_circles possui um formato ainda mais difícil de separar utilizando uma linha reta.

Por isso, o kernel linear apresentou um resultado inferior.

O kernel rbf apresentou um resultado melhor porque consegue trabalhar com dados que possuem uma separação não linear.

A alteração de C também modificou a capacidade do modelo de separar os dados. Já o aumento de noise tornou a classificação mais difícil, pois os pontos ficaram mais misturados.

Assim como no make_moons, o rbf foi o kernel que melhor se adaptou ao formato dos dados.

```python
# Exercício 5.

import numpy as np
from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

valores_noise = [0.10, 0.15, 0.20, 0.25]

for noise in valores_noise:

    X, y = make_circles(
        n_samples=500,
        noise=noise,
        factor=0.4,
        random_state=0
    )

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=0
    )

    modelo = SVC(kernel='rbf', C=10)
    modelo.fit(X_train, y_train)

    print(
        "Noise =", noise,
        "- Acurácia =", modelo.score(X_test, y_test)
    )

# Alterando somente o factor
valores_factor = [0.3, 0.35, 0.4, 0.45, 0.5]

for factor in valores_factor:

    X, y = make_circles(
        n_samples=500,
        noise=0.15,
        factor=factor,
        random_state=0
    )

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, random_state=0
    )

    modelo = SVC(kernel='rbf', C=10)
    modelo.fit(X_train, y_train)

    print(
        "Factor =", factor,
        "- Acurácia =", modelo.score(X_test, y_test)
    )
```

Ao modificar o noise, percebemos que o aumento do ruído deixa os dados mais desorganizados. Com isso, o SVM passa a ter mais dificuldade para encontrar uma separação adequada entre as classes.

Já o factor controla o tamanho do círculo interno em relação ao círculo externo.

Quando o factor muda, a distância entre os dois grupos também muda. Dependendo do valor utilizado, os círculos podem ficar mais próximos ou mais afastados.

Nesse caso, o rbf consegue lidar melhor com essas mudanças porque foi desenvolvido para trabalhar com separações não lineares.

As melhores acurácias foram obtidas principalmente quando os dados apresentavam uma separação mais clara entre as classes e quando foi utilizado o kernel rbf.

Também foi possível perceber que valores muito altos de noise prejudicam a classificação, pois deixam os dados mais misturados. O parâmetro C também influencia o resultado, sendo necessário encontrar um valor adequado para evitar tanto erros de classificação quanto overfitting.

Portanto, de maneira geral, os melhores resultados foram obtidos com o kernel RBF, valores adequados de C e conjuntos de dados com pouco ruído.
