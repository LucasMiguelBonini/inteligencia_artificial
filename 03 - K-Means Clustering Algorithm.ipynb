Exercício:

1. Gerar 10 novos bancos de dados utilizando comando make_blobs fazendo 10 aranjos diferentes, alterando os parâmetros: `n_samples` = 100, 800 e 1500, `centers` = 3, 5 e 8, `cluster_std` = 0.40, 0.70 e 0.90, e `random_state` = 10, 20 e 30. Para cada arranjo  dos parâmetros modificados verificar e anotar a eficiência do método para predição dos centróides. Comentar os resultados.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
```

```python
X, y_true, centers_true = make_blobs(
    n_samples=100,
    centers=3,
    cluster_std=0.40,
    random_state=10,
    return_centers=True
)

kmeans = KMeans(n_clusters=3, n_init=10)
kmeans.fit(X)

print("Experimento 1")
print("Centróides reais:")
print(centers_true)

print("\nCentróides encontrados pelo K-Means:")
print(kmeans.cluster_centers_)

plt.scatter(X[:, 0], X[:, 1], c=kmeans.labels_, s=20)
plt.scatter(centers_true[:, 0], centers_true[:, 1], c='red', s=100)

plt.title("Experimento 1")
plt.show()
```
