````python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report
````

````python
print("Percentual de fraude por tipo de vinho:")

for tipo in df['type'].unique():
    dados = df[df['type'] == tipo]

    percentual = (dados['quality'] == 'fraud').mean() * 100

    print(f"{tipo}: {percentual:.2f}%")
````

````python
df = pd.get_dummies(df, columns=['type'], drop_first=True)
````

````python
df.head()
````

````python
X = df.drop('quality', axis=1)

y = df['quality']
````

````python
print("Variáveis de entrada:")
print(X.columns)

print("\nVariável que será prevista:")
print(y.name)
````

````python
from sklearn.model_selection import train_test_split
````

````python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)
````

````python
print("Treinamento:", X_train.shape)
print("Teste:", X_test.shape)
````

````python
from sklearn.preprocessing import StandardScaler
````

````python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
````

````python
from sklearn.svm import SVC
````

````python
model = SVC()

model.fit(X_train, y_train)
````

````python
y_pred = model.predict(X_test)
````

````python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)
````

````python
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(y_test, y_pred, pos_label='Fraud')
recall = recall_score(y_test, y_pred, pos_label='Fraud')
f1 = f1_score(y_test, y_pred, pos_label='Fraud')

print(f"Acurácia: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")
````

````python
param_grid = {
    'kernel': ['linear', 'rbf', 'sigmoid'],
    'C': [0.01, 0.1, 0.5, 1, 2],
    'gamma': ['scale', 'auto']
}
````

````python
from sklearn.metrics import make_scorer, f1_score

f1_scorer = make_scorer(f1_score, pos_label='Fraud')

grid = GridSearchCV(
    SVC(),
    param_grid,
    cv=5,
    scoring=f1_scorer
)

grid.fit(X_train, y_train)
````

````python
print("Melhores parâmetros:")
print(grid.best_params_)

print("\nMelhor F1-Score:")
print(grid.best_score_)
````

````python
best_model = grid.best_estimator_

y_pred = best_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, pos_label='Fraud')
recall = recall_score(y_test, y_pred, pos_label='Fraud')
f1 = f1_score(y_test, y_pred, pos_label='Fraud')

print(f"Acurácia: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")
````

````python
print(classification_report(y_test, y_pred))
````

````python
resultados = pd.DataFrame(grid.cv_results_)

resultados = resultados[
    [
        'param_kernel',
        'param_C',
        'param_gamma',
        'mean_test_score',
        'rank_test_score'
    ]
]

resultados = resultados.sort_values('rank_test_score')

resultados
````

````python
melhor = resultados.iloc[0]

print("Melhor configuração encontrada:")
print(f"Kernel: {melhor['param_kernel']}")
print(f"C: {melhor['param_C']}")
print(f"Gamma: {melhor['param_gamma']}")
print(f"F1-Score: {melhor['mean_test_score']:.4f}")
````

### Conclusão

Neste exercício foi utilizado o algoritmo SVM (Support Vector Machine) para classificar amostras de vinho entre fraudadas e legítimas. Inicialmente, foi realizada uma análise do dataset para verificar as classes existentes e a distribuição dos tipos de vinho, considerando vinhos tintos e brancos.

Para utilizar o SVM, a variável categórica referente ao tipo de vinho foi convertida para um formato numérico. Em seguida, os dados foram separados em 70% para treinamento e 30% para teste. Também foi aplicado o Standard Scale para padronizar os valores das variáveis antes do treinamento do modelo.

Foram testadas diferentes combinações dos hiperparâmetros kernel, C e gamma, utilizando os kernels linear, rbf e sigmoid. Ao todo, foram avaliadas 30 combinações diferentes por meio do GridSearchCV.

O melhor resultado encontrado foi obtido com o kernel sigmoid, utilizando C = 2.0 e gamma = auto, apresentando um F1-Score de 0.1082. O resultado mostra que, apesar de o modelo conseguir realizar a classificação, ele apresentou dificuldade para identificar corretamente as amostras fraudulentas.

Isso pode estar relacionado principalmente ao desbalanceamento entre as classes de vinhos fraudados e legítimos. Em um problema como esse, apenas a acurácia não é suficiente para avaliar o modelo, sendo importante também observar métricas como precision, recall e F1-Score, principalmente para verificar a capacidade do modelo de identificar as fraudes.

Dessa forma, o exercício mostrou na prática como o SVM pode ser utilizado em um problema de classificação e como a escolha dos hiperparâmetros pode influenciar seu desempenho. Também foi possível perceber a importância de analisar a distribuição dos dados e utilizar métricas adequadas quando existem diferenças significativas entre as classes.
