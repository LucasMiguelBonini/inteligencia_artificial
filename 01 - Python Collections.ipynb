```python
# 4.Escreva uma função em Python que recebe duas listas e retorna True se elas tiverem pelo menos um membro em comum.

def elemento_comum(lista1, lista2):
  for i in lista1:
    if i in lista2:
      return True
  return False

lista1 = [10, 20, 30, 40]
lista2 = [50, 60, 70, 10]

print(elemento_comum(lista1, lista2))
```

```python
# 5.Escreva um programa em Python para gerar e imprimir uma lista dos primeiros e últimos 5 elementos, onde os valores são números quadrados entre 1 e 30 (ambos incluídos).

squares = []

for i in range(1, 31):
  squares.append(i**2)

print("Primeiros 5 elementos:", squares[:5])
print("Últimos 5 elementos:", squares[-5:])
```

```python
# 6.Escreva um programa em Python para verificar se cada número é primo em uma determinada lista de números. Retorne True se todos os números forem primos, caso contrário, False.

def primo(numero):
  if numero < 2:
    return False

  for i in range (2, numero):
    if numero % i == 0:
      return False

  return True

def todos_primos(lista):
  for numero in lista:
    if primo(numero) == False:
      return False
  return True

print(todos_primos([0, 3, 4, 7, 9]))
print(todos_primos([3, 5, 7, 13]))
print(todos_primos ([1, 5, 3]))
```

```python
# 9.Escreva um programa em Python para encontrar itens repetidos em uma tupla.

tupla = (0, 0, 2, 3, 4, 5, 5, 6, 7)

repetidos = set()

for item in tupla:
    if tupla.count(item) > 1:
        repetidos.add(item)

print("Itens repetidos:", repetidos)
```

```python
# 15.Escreva um script Python para gerar e imprimir um dicionário que contém um número (entre 1 e n) na forma (x, x*x). Dicionário de amostra: (n = 5). Saída esperada: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}.

n = 10

dic = {}

for x in range(1, n + 1):
    dic[x] = x * x

print(dic)
```

```python
# 16.Escreva um programa em Python para obter os valores máximo e mínimo de um dicionário.

dic = {
    "a": 3,
    "b": 22,
    "c": 8,
    "d": 33,
    "e": 77
}

print("Valor máximo:", max(dic.values()))
print("Valor mínimo:", min(dic.values()))
```

```python
""" 17.Escreva um programa em Python para imprimir todos os valores distintos em um dicionário. Dados de amostra:
 [{"V":"S001"}, {"V": "S002"}, {"VI": "S001"}, {"VI": "S005"}, {"VII":"S005"}, {"V":"S009"},{"VIII":"S007"}].
 Saída esperada: Valores exclusivos:{'S005', 'S002', 'S007', 'S001', 'S009'}."""

dic = {
    "V": "S001",
    "VI": "S002",
    "VII": "S001",
    "VIII": "S005",
    "IX": "S005",
    "X": "S009",
    "XI": "S007"
}

valores = set(dic.values())

print("Valores exclusivos:", valores)
```

```python
"""18.Escreva um programa em Python para filtrar números pares de um dicionário de valores. Dicionário original: {'V': [1, 4, 6, 10], 'VI': [1, 4, 12], 'VII': [1, 3, 8]}.
Filtrar números pares dos valores do dicionário: {'V': [4, 6, 10], 'VI': [4, 12], 'VII': [8]}.
Dicionário original: {'V': [1, 3, 5], 'VI': [1, 5], 'VII': [2, 7, 9]}.
Filtrar números pares dos valores do dicionário: {'V': [], 'VI': [], 'VII': [2]}."""

dic = {
    "V": [1, 4, 6, 10],
    "VI": [1, 4, 12],
    "VII": [1, 3, 8]
}

novo_dic = {}

for chave in dic:
    novo_dic[chave] = []

    for numero in dic[chave]:
        if numero % 2 == 0:
            novo_dic[chave].append(numero)

print("Dicionário original:", dic)
print("Filtrar números pares dos valores do dicionário:", novo_dic)
```
